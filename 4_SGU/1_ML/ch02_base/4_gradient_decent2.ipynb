```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 定义目标函数
def J(x):
    return (x**2-2)**2
# 定义梯度函数（导函数）
def gradient(x):
    return 4*x**3-8*x
```

```python
# 用列表保存点的变化轨迹
y_list = []
x_list = []
```

```python
# 定义超参数和x的初始值
alpha = 0.1
x = 1
```

```python
# 重复迭代100次
while np.abs(grad := gradient(x)) > 1e-10:
    y = J(x)
    # 记录当前点的坐标
    x_list.append(x)
    y_list.append(y)
    print(f'x:{x}, y:{y}')
    # 计算梯度
    # 根据公式更新参数
    x = x - alpha * grad
print(len(x_list))
```

```python
# 画出图像
# x = np.arange(5,1,0.01) # 图像不明显
# x = np.arange(1,1,0.01) # 图像不明显
x = np.arange(1.414,1.425,0.01)
plt.plot(x,J(x))
plt.plot(x_list,y_list,c="r")
plt.scatter(x_list,y_list,c="r")
plt.show()
```
