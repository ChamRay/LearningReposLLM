```python
import torch
```

```python
tensor1 = torch.randint(1,10,(3,2,4))
print(tensor1)
```

```python
print(tensor1.sum())
print(tensor1.sum(dim=0)) # 求和第0维度，降维操作
```

```python
print(tensor1.mean(dtype=float))
print(tensor1.mean(dim=2,dtype=float))
```

```python
# 求标准方差
tensor1 = tensor1.float()
print(tensor1.std())
```

```python
# max min
print(tensor1.max())
print(tensor1.min())
```

```python
print(tensor1.min(dim=2))
```

```python
print(tensor1.sort())
```
