```python
import torch
```

```python
# 1. 四则运算
tensor1 = torch.randint(low=0, high=10, size=(2,3))
print(tensor1)
```

```python
print(tensor1+10)
print(tensor1)
```

```python
print(tensor1.add(10))
print(tensor1)
```

```python
print(tensor1.add_(10))
print(tensor1)
```

```python
# 求幂
print(tensor1.sub_(10))
print(tensor1**2)
print(tensor1.pow(2))
print(tensor1.pow_(2))
```

```python
print(tensor1.sqrt())
tensor1 = tensor1.float() # 注意，sqrt_默认会转成float，要提前转成浮点类型
print(tensor1.sqrt_())
```

```python
# 以e为底计算
tensor1 = torch.tensor([1.0,2.0,3.0])
print(tensor1)
print(2.7183**tensor1)
print(tensor1.exp())
```

```python
# 计算log
print(tensor1.log())
```

```python
# 哈达玛积 元素级乘法
tensor1 = torch.tensor([[1,2],[3,4]])
tensor2 = torch.tensor([[1,2],[3,4]])
print(tensor1+tensor2)
print(tensor1*tensor2)
print(tensor1.mul(tensor2))
```

```python
# 张量乘法运算
# mm()严格用于二维矩阵
tensor1 = torch.tensor([[1,2,3],[3,4,5]])
tensor2 = torch.tensor([[1,2],[3,4],[5,6]])
print(tensor1)
print(tensor2)
print(tensor1.mm(tensor2))
print(tensor1 @ tensor2)
print(tensor1.matmul(tensor2))
```
