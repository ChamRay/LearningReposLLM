```python
import torch
```

```python
x = torch.tensor(2.0, requires_grad=True)
y = x.detach()
print(x)
print(y)
```

```python
print(x.requires_grad)
print(y.requires_grad)
```

```python
print(id(x))
print(id(y))
```

```python
print(x.untyped_storage().data_ptr())
print(y.untyped_storage().data_ptr())
```

```python
# print(x)
# y.zero_()
# print(x)
```

```python
z1 = x ** 2
z2 = y ** 2
print(z1)
print(z2)
```

```python
print(z1.requires_grad)
print(z2.requires_grad)
```

```python
z1.sum().backward()
```

```python
x = torch.ones(2,2, requires_grad=True)
y = x * x
print(x)
print(y)
```

```python
u = y.detach()
print(u)
```

```python
# 让u参与新张量的计算
z = u * x
```

```python
z.sum().backward()
print(x.grad)
```
