```python
import torch
```

```python
tensor1 = torch.randint(1,10,(2,3,6))
print(tensor1)
```

1. 交换维度

```python
# 1. 交换两个维度
print(tensor1.transpose(1,2))
print(tensor1.transpose(0,2))
```

```python
# 2. 重新排列多个维度
print(tensor1.permute(1,0,2))
```

2. 调整形状

```python
tensor1 = torch.randint(1,10,(3,5,4))
print(tensor1.shape)
```

```python
# 1. reshape
print(tensor1.reshape(6,10))
```

```python
print(tensor1.reshape(3,-1)) # 3行，列自动算出来
```

```python
# 2 view
print(tensor1.is_contiguous())
print(tensor1.view(-1,10))
```

```python
print(tensor1.shape)
```

```python
print(tensor1.shape)
tensor1 = tensor1.permute(2,1,0)
print(tensor1.shape)
print(tensor1.is_contiguous())
tensor1 = tensor1.contiguous()
print(tensor1.is_contiguous())
```

3. 增加或删除维度

```python
tensor1 = torch.tensor([1,2,3])
print(tensor1)
```

```python
# 在0维上增加一个维度 dim=0或者-2
print(tensor1.unsqueeze(dim=0))
```

```python
# 在1维上增加一个维度 dim=1或者-1
print(tensor1.unsqueeze(dim=1))
```

```python
# 减少维度
print(tensor1.squeeze(dim=0))
```

4. 张量拼接操作

```python
tensor1 = torch.randint(1,10,(2,2,5))
tensor2 = torch.randint(1,10,(2,3,5))
print(tensor1)
print(tensor2)
```

```python
# 张量拼接
print(torch.cat((tensor1,tensor2),dim=1))
```

```python
# 张量堆叠
tensor1 = torch.randint(1,10,(4,3,5))
tensor2 = torch.randint(1,10,(4,3,5))
print(tensor1)
print(tensor2)
```

```python
tensor3 = torch.stack([tensor1,tensor2],dim=0)
print(tensor3.shape)
print(tensor3)
```
