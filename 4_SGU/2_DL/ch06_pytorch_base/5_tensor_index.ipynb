```python
import torch
```

```python
# 简单索引操作
tensor1 = torch.randint(1,9,(3,5,4))
print(tensor1)
print(tensor1.shape)
print()
```

```python
print(tensor1[2,1,3])
```

```python
print(tensor1[2,3])
```

```python
print(tensor1[:,3]) # 取所有矩阵的第3行
```

```python
print(tensor1[:,3,2])
```

```python
print(tensor1[1:])
```

```python
mask = tensor1[:,:,0]>5 #行首元素大于5
print(mask)
```

```python
print(tensor1[mask]) # 筛选所有true的行
```
