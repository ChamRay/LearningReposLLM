```python
import torch
import numpy as np
```

1. 基本张量创建 （1）按内容创建张量

```python
tensor1 = torch.tensor(10)
print(tensor1)
print(tensor1.size()) # 0维，显示[]
print(tensor1.dtype) # 默认int64
```

```python
tensor2 = torch.tensor([1.0, 2, 3])
print(tensor2)
print(tensor2.size())
print(tensor2.dtype)
```

```python
# 传入一个二维的ndarray
tensor3 = torch.tensor(np.array([[1,2,3],[4,5,6]]))
print(tensor3)
print(tensor3.size())
print(tensor3.dtype)
```

(2)创建指定形状的张量

```python
tensor4 = torch.Tensor(3,2,4)
print(tensor4)
print(tensor4.size())
print(tensor4.dtype)
```

```python
tensor5 = torch.Tensor([[1,2,3],[4,5,6]])
print(tensor5)
print(tensor5.size())
print(tensor5.dtype) # 不管传入什么样的数据，都会被转为float32
```

(3)创建指定类型的张量

```python
tensor6 = torch.IntTensor([[1,2,3],[4,5,6]])
tensor7 = torch.tensor([[1,2,3],[4,5,6]],dtype=torch.int64)
print(tensor6.dtype)
print(tensor7.dtype)
```

2. 指定区间的张量创建

```python
tensor1 = torch.arange(10,30,2) # 按步长2生成元素
print(tensor1)
```

```python
tensor2 = torch.arange(6)
print(tensor2)
```

```python
tensor3 = torch.linspace(10,30,5) # 生成5个元素
print(tensor3)
```

```python
tensor4 = torch.logspace(10,30,5,2) # 生成5个按底的指数
print(tensor4)
```

3. 按数值填充张量

```python
tensor1 = torch.zeros(2,3)
print(tensor1)
```

```python
tensor2 = torch.ones_like(tensor1)
print(tensor2)
```

```python
tensor3 = torch.full((3,2),6)
print(tensor3)
print(tensor3.dtype)
```

```python
tensor4 = torch.empty_like(tensor3)
print(tensor4)
```

```python
# 单位矩阵
tensor5 = torch.eye(3)
print(tensor5)
```

4. 创建随机张量

```python
tensor1 = torch.rand(2,3) # 均匀分布
print(tensor1)
```

```python
tensor2 = torch.rand_like(tensor1)
print(tensor2)
```

```python
tensor3 = torch.randint(low=0,high=100,size=(2,3))
print(tensor3)
```

```python
tensor4 = torch.randint_like(tensor3,low=0,high=20)
print(tensor4)
```

```python
tensor5 = torch.randn(2,3)
print(tensor5)
```

```python
tensor6 = torch.randn_like(tensor5)
print(tensor6)
```

```python
tensor7 = torch.normal(5,2,tensor6.shape)
print(tensor7)
```

```python
tensor1 = torch.randperm(10) # 随机重新排列
print(tensor1)
```

```python
# 查看和指定随机数种子
print(torch.random.initial_seed())
torch.manual_seed(42)
print(torch.random.initial_seed())
```
