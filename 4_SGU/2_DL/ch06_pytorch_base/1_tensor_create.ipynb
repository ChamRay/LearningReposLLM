```python
import torch
import numpy as np
```

1. 基本张量创建 （1）按内容创建张量

```python
tensor1 = torch.tensor(10)
print(tensor1)
print(tensor1.size()) # 0维，显示[]
print(tensor1.dtype) # 默认int64
```

```python
tensor2 = torch.tensor([1.0, 2, 3])
print(tensor2)
print(tensor2.size())
print(tensor2.dtype)
```

```python
# 传入一个二维的ndarray
tensor3 = torch.tensor(np.array([[1,2,3],[4,5,6]]))
print(tensor3)
print(tensor3.size())
print(tensor3.dtype)
```

(2)创建指定形状的张量

```python
tensor4 = torch.Tensor(3,2,4)
print(tensor4)
print(tensor4.size())
print(tensor4.dtype)
```

```python
tensor5 = torch.Tensor([[1,2,3],[4,5,6]])
print(tensor5)
print(tensor5.size())
print(tensor5.dtype) # 不管传入什么样的数据，都会被转为float32
```

(3)创建指定类型的张量

```python
tensor6 = torch.IntTensor([[1,2,3],[4,5,6]])
tensor7 = torch.tensor([[1,2,3],[4,5,6]],dtype=torch.int64)
print(tensor6.dtype)
print(tensor7.dtype)
```
