```python
import torch
```

```python
x = torch.tensor(10.0)
y = torch.tensor(2.0)
print(x)
print(y)
```

```python
w = torch.rand(1,1,requires_grad=True)
b = torch.rand(1,1,requires_grad=True)
print(w)
print(b)
```

```python
z = w * x + b
print(z)
```

```python
print(x.is_leaf)
print(w.is_leaf)
print(b.is_leaf)
print(y.is_leaf)
print(z.is_leaf)
```

```python
loss = torch.nn.MSELoss()
loss_value = loss(z,y)
print(loss_value)
print(loss_value.is_leaf)
```

```python
loss_value.backward()
```

```python
print(w.grad)
print(b.grad)
```
