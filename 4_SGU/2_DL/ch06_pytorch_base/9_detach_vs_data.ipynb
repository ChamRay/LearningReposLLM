```python
import torch
```

```python
x1 = torch.tensor([1.0,2,3],requires_grad=True)
x2 = torch.tensor([1.0,2,3],requires_grad=True)
print(x1)
print(x2)
```

```python
y1 = x1.sigmoid()
y2 = x2.sigmoid()
print(y1)
print(y2)
```

```python
# y1.sum().backward()
# print(x1.grad)
# y2.sum().backward()
# print(x2.grad)
```

```python
z1 = y1.data
z2 = y2.detach()
print(z1)
print(z2)
```

```python
print(z1.requires_grad)
print(z2.requires_grad)
```

```python
z1.zero_()
z2.zero_()
print(y1)
print(y2)
```

```python
y1.sum().backward()
print(x1.grad)
y2.sum().backward()
print(x2.grad)

# 相比较之下，自动微分就是能在合适的时候抛出错误，而data会使用默认值0，让错误继续执行下去。
```
