```python
import torch
import numpy as np
```

1. 张量元素类型转换

```python
tensor1 = torch.tensor([1, 2, 3])
print(tensor1.dtype)
```

```python
tensor2 = tensor1.type(torch.float32)
print(tensor2.dtype)
```

```python
tensor2 = tensor1.half()
print(tensor2.dtype)
```

```python
# 转成复数
tensor1 = tensor1.to(torch.complex64)
print(tensor1,tensor1.dtype)
```

2. tensor转成ndarray

```python
np.set_printoptions(precision=4)
tensor1 = torch.rand(3,2)
print(tensor1)
```

```python
ndarray2 = tensor1.numpy().copy()
print(ndarray2)
```

```python
ndarray1 = tensor1.numpy()
print(ndarray1)
```

```python
tensor1[:,0] = 10
print(tensor1)
print(ndarray1)
print(ndarray2)
```

```python
# 将ndarray转成tensor
ndarray1 = np.random.randn(3,2)
tensor1 = torch.from_numpy(ndarray1)
print(tensor1)
print(ndarray1) # 默认都是内存共享的
```

```python
# 非内存共享转换，通过创建tensor
ndarray1 = np.random.randn(3,2)
tensor1 = torch.tensor(ndarray1)
print(tensor1)
print(ndarray1)
```

3. tensor和标量进行转换

```python
tensor1 = torch.tensor(10)# 标量->tensor
print(tensor1)
```

```python
print(tensor1.item())# tensor->标量
```
