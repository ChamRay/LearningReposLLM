```python
import torch
import torch.nn as nn
import jieba
```

```python
text = "自然语言是由文字构成的，而语言的含义是由单词构成的。即单词是含义的最小单位。因此为了让计算机理解自然语言，首先要让它理解单词含义。"
```

```python
# 1. 分词
original_words = jieba.lcut(text)
print(original_words)
```

```python
# 自定义一组停用词
stop_words = {"的","是","而","由","，","。","、"}
```

```python
# 2. 过滤停用词和标点符号，构建词表
words = [word for word in original_words if word not in stop_words]
print(words)
```

```python
# 3. 构建词表
id2word = list(set(words))
print(id2word)
```

```python
# 4. 构建字典，保存word到索引号的映射
word2id = dict()
for id,word in enumerate(id2word):
    word2id[word] = id
print(word2id)
```

```python
# 5. 构建一个嵌入层
embed = nn.Embedding(num_embeddings=len(word2id), embedding_dim=5)
```

```python
# 6. 前向传播，代入单词表引号，得到词向量
for id,word in enumerate(id2word):
    word_vec = embed(torch.tensor(id))
    print(f"{id:>2}:{word:8}\t{word_vec.detach().numpy()}")
```
