```python
import torch
import torch.nn as nn
```

```python
# 定义RNN层
rnn = nn.RNN(input_size=8,hidden_size=16,num_layers=2)
```

```python
# 定义输入数据和初始隐藏状态
input = torch.randn(1,3,8) # seq_len 序列长度 batch_size 批量大小 input_size 和RNN保持一致
hidden = torch.randn(2,3,16) # num_layers 初始隐藏状态，和RNN中保持一致 batch_size 和input保持一致 hidden_size 和RNN中保持一致
```

```python
# 前向传播，得到输出和隐状态
out,hn = rnn(input,hidden)
print(out)
print(hn) # 相当于是个快照
print(out.shape)
print(hn.shape)
```
