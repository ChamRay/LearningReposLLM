```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

```python
# 1. 读取图片
img = plt.imread("../data/duck.jpg")
print(img.shape)
```

```python
# 2. 将图片数据调整为卷积层输入特征图对应的形状
input = torch.tensor(img).permute(2,0,1).float()
print("输入的特征图形状为：",input.size())
```

```python
conv = nn.Conv2d(in_channels=3,out_channels=3,kernel_size=9,stride=3,padding=0,bias=False)
```

```python
# 4. 前向传播，将卷积层应用到输入特征图上
output = conv(input)
print("输出特征图的形状：",output.size())
```

```python
# 5. 定义池化层
pool = nn.MaxPool2d(kernel_size=6,stride=6,padding=1,)
```

```python
# 6. 前向传播，进行池化操作
output2 = pool(output)
print(output2.size())
```

```python
# 画图
output = torch.clamp(output.int(),0,255)
output = torch.permute(output,(1,2,0)).detach().numpy()
output2 = torch.clamp(output2.int(),0,255)
output2 = torch.permute(output2,(1,2,0)).detach().numpy()

fig,ax = plt.subplots(1,3,figsize=(15,5))
ax[0].imshow(img)
ax[1].imshow(output)
ax[2].imshow(output2)
plt.show()
```
