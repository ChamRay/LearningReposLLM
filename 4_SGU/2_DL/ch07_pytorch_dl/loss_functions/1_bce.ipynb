```python
import torch
import torch.nn as nn
```

```python
input = torch.randn(3,2)
print(input)
# 得到预测概率值
pred = torch.sigmoid(input)
print(pred)
```

```python
target = torch.tensor([[0,1],[1,0],[0,1]],dtype=torch.float32)
print(target)
```

```python
loss = nn.BCELoss()
print(loss(pred, target))
```
