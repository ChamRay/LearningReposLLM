```python
import torch
import torch.nn as nn
```

```python
# 定义数据
input = torch.randn(3,5)
target = torch.randn(3,5)
print(input)
print(target)
```

```python
# MSE
mse_loss = nn.MSELoss()
print(mse_loss(input, target))
```

```python
# MAE
mae_loss = nn.L1Loss()
print(mae_loss(input, target))
```

```python
# Smooth L1
smooth = nn.SmoothL1Loss()
print(smooth(input, target))
```
