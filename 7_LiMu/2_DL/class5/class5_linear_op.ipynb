```python
import torch
# 标量由只有一个元素的张量表示
x = torch.tensor([3.0])
y = torch.tensor([2.0])
x + y,x*y,x/y,x**y

```

```python
# 可以将向量视作标量值组成的列表
x = torch.arange(4)
x
```

```python
# 通过张量的索引来访问任意元素
x[3]
```

```python
# 访问张量的长度
print(len(x))
# 张量的形状
print(x.shape)
```

```python
# 通过指定两个分量m和n来创建一个形状为mxn的矩阵
A = torch.arange(20).reshape(5,4)
A
```

```python
# 矩阵的转置
A.T
```

```python
# 对称矩阵,其转置等于自己
B = torch.tensor([[1,2,3],[2,0,4],[3,4,5]])

B,B.T==B
```

```python
# 就像向量是标量的推广,矩阵是向量的推广一样,可以构建具有更多轴的数据结构
X = torch.arange(24).reshape(2,3,4)
X
```

```python
# 给定具有相同形状的任何两个张量,任何暗元素二元运算的结果都将是相同形状的张量
A = torch.arange(20,dtype=torch.float32).reshape(5,4)
B = A.clone() # 通过分配新内存,将A的一个副本分配给B

A,A+B
```

```python
# 两个矩阵的按元素乘法称为 哈达玛积
A * B
```

```python
# 标量与矩阵的运算
a = 2
X = torch.arange(24).reshape(2,3,4)
a+X,(a*X).shape
```

```python
# 计算器元素的和
x = torch.arange(4,dtype=torch.float32)

x , x.sum()
```

```python
# 表示任意形状张量的元素和
A = torch.arange(20*2).reshape(2,5,4)
A,A.shape,A.sum()
```

```python
# 指定求和张量的轴/维度
A_sum_axis0 = A.sum(dim=0)
A_sum_axis0,A_sum_axis0.shape
```

```python
A_sum_axis1 = A.sum(dim=1)
A_sum_axis1,A_sum_axis1.shape
```

```python
A_sum_axis2 = A.sum(dim=2)
A_sum_axis2,A_sum_axis2.shape
```

```python
A = torch.arange(20,dtype=torch.float32).reshape(5,4)
A.shape,A.sum()
```

```python
# 求和相关的量是平均值
A.mean(dtype=torch.float32),A.sum()/A.numel()
```

```python
# 按照某个维度算均值
A.mean(dim=0,dtype=torch.float32),A.sum(dim=0,dtype=torch.float32)/A.shape[0]
```

```python
# 求和,但是不丢失维度
sum_A = A.sum(dim=1,keepdim=True)
sum_A
```

```python
# 通过广播将A除以sum_A
A / sum_A
```

```python
# 某个维度计算A元素的累加求和
A.cumsum(dim=0)
```

```python
# 点积是相同位置的按元素乘积的和
x = torch.arange(4,dtype=torch.float32)
y = torch.ones(4,dtype=torch.float32)
x , y , torch.dot(x,y)
```

```python
# 通过执行案元素乘法,然后进行求和表示两个向量的点积
torch.sum(x * y)
```

```python
# x = x.to(torch.long)
# 矩阵向量Ax是一个长度为m的列向量,其i^th元素是点积a_i^TX
# A = torch.tensor([[1,1,2]])
# x = torch.tensor([1,2,3])
print(A.shape)
print(x.shape)
torch.mv(A,x)
```

```python
# 可以将矩阵-矩阵乘法AB看做是简单地执行m次矩阵-向量积,并将结果拼接在一起,形成一个nxm矩阵
B = torch.ones(4,3,dtype=torch.float32)
A = A.to(torch.float32)
print(A.shape)
print(B.shape)
torch.mm(A,B)
```

```python
# L2范数,是向量元素平方和的平方根
u = torch.tensor([3.0,-4.0])
torch.norm(u)
```

```python
# L1范数,是向量元素的绝对值之和
torch.abs(u).sum()
```

```python
# 矩阵的 弗罗贝尼乌斯范数 是矩阵元素的平方和的平方根
torch.norm(torch.ones(4,9))
```
