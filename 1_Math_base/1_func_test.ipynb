```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0, 6, 0.1)
y = np.sin(x)
```

```python
print(x)
print(y)
```

```python
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = sin(x)")
plt.show()
```

```python
# 导函数 y'=cos(x)
y1 = np.cos(x)
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y' = cos(x)")
plt.show()
```

```python
# 将函数和导函数图像放在一张图中
plt.plot(x,y,label="y=sin(x)")
plt.plot(x,y1,label="y'=cos(x)",linestyle="--")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
